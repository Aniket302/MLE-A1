```python
import os
import pandas as pd

# Folder containing the CSV files
folder_path = 'data/'

# List all CSV files in the folder
csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]

# Loop through each file and extract useful information
for file in csv_files:
    file_path = os.path.join(folder_path, file)
    
    # Load the CSV file into a pandas DataFrame
    df = pd.read_csv(file_path)
    
    # Get the file size
    file_size = os.path.getsize(file_path)  # in bytes
    
    # Get column names
    columns = df.columns.tolist()
    
    # Get the shape of the DataFrame
    num_rows, num_columns = df.shape
    
    # Get data types of each column
    column_types = df.dtypes.to_dict()
    
    # Count missing values
    missing_values = df.isnull().sum().to_dict()
    
    # Descriptive statistics for numerical columns
    descriptive_stats = df.describe()
    
    # Get the first two rows
    first_two_rows = df.head(2)
    
    # Print the file information
    print(f"File: {file}")
    print(f"Columns: {columns}")
    print(f"Shape (Rows x Columns): {num_rows} x {num_columns}")
    print(f"Data Types: {column_types}")
    print(f"Missing Values: {missing_values}")
    # print(f"Descriptive Statistics:\n{descriptive_stats}")
    print(f"First Two Rows:\n{first_two_rows}")
    print("-" * 50)
```
